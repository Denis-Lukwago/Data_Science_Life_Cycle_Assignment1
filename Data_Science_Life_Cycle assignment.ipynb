```python
#Handling missing values
#Filling missing values (NaN) in the stroke column with the median
automobiles["stroke"] = automobiles["stroke"].fillna(automobiles["stroke"].median())
#Filling the missing values in horsepower-binned with its mode
automobiles["horsepower-binned"] = automobiles["horsepower-binned"].fillna(automobiles["horsepower-binned"].mode()[0])
automobiles.isnull().sum()
```

```python
#Drive Wheels and Body Style
pivot = pd.pivot_table(
    automobiles,
    values="price",
    index="drive-wheels",
    columns="body-style",
    aggfunc="mean"
)

sns.heatmap(pivot, annot=True)
#RWD convertibles and hardtops tend to have the highest average prices.
#FWD hatchbacks generally have the lowest average prices.
```

Examining relationship between price and other variables

```python
corr_matrix = automobiles.corr(numeric_only=True)

corr_matrix["price"].sort_values(
    ascending=False
)
```

Top predictors are; engine size, curb weight, horsepower, city-L/100km, width and length
These variables consistently show the strongest relationship with price.

Recommendations
Based on the analysis, vehicle pricing should focus primarily on engine size, horsepower, and curb weight because they exhibit the strongest relationship with price. 
Rear-wheel-drive vehicles, hardtops, and convertibles can be priced at a premium due to their higher average market value. 
Fuel-efficient vehicles generally occupy lower price ranges, while larger vehicles with powerful engines command higher prices. 
This implies that dealers should prioritize these key features when establishing pricing models and forecasting vehicle values.
